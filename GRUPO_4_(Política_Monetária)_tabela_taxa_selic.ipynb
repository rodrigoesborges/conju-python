```python

import pandas as pd
from bcb import sgs

# Obter os dados da Selic Meta (código 432)
selic = sgs.get(432, start="2023-01-01")

# Organizar os dados por mês
selic_mensal = selic.resample("MS").last()

# Criar a tabela base em formato longo
tabela_base = selic_mensal.reset_index()

tabela_base.columns = ["periodo", "valor"]

tabela_base["indicador"] = "Selic Meta"

tabela_base = tabela_base[
    ["periodo", "indicador", "valor"]
]

# Exibir a tabela
display(tabela_base.tail(12))
```

```python

# Tabela Intermediária 1 - Selic vigente no ano

tabela_intermediaria_1 = tabela_base.copy()

tabela_intermediaria_1["ano"] = (
    tabela_intermediaria_1["periodo"].dt.year
)

tabela_intermediaria_1 = (
    tabela_intermediaria_1
    .sort_values("periodo")
    .reset_index(drop=True)
)

display(tabela_intermediaria_1.tail(12))
```

```python

# Tabela Intermediária 2 - Selic vigente e comparação anual

tabela_intermediaria_2 = tabela_base.copy()

# Organizar por data
tabela_intermediaria_2 = (
    tabela_intermediaria_2
    .sort_values(["indicador", "periodo"])
    .reset_index(drop=True)
)

# Taxa vigente 12 meses antes
tabela_intermediaria_2["valor_12m_antes"] = (
    tabela_intermediaria_2
    .groupby("indicador")["valor"]
    .shift(12)
)

# Diferença em pontos percentuais
tabela_intermediaria_2["variacao_12m_pp"] = (
    tabela_intermediaria_2["valor"]
    - tabela_intermediaria_2["valor_12m_antes"]
)

display(tabela_intermediaria_2.tail(12))
```

```python

# Tabela Intermediária 3 - Comparação anual da Selic

tabela_intermediaria_3 = tabela_base.copy()

# Criar a coluna com a data de 12 meses antes
tabela_intermediaria_3["periodo_anterior"] = (
    tabela_intermediaria_3["periodo"]
    - pd.DateOffset(years=1)
)

# Preparar os valores do ano anterior
tabela_anterior = tabela_base.rename(
    columns={
        "periodo": "periodo_anterior",
        "valor": "valor_anterior"
    }
)

# Juntar as tabelas
tabela_intermediaria_3 = tabela_intermediaria_3.merge(
    tabela_anterior,
    on=["indicador", "periodo_anterior"],
    how="left"
)

# Calcular a variação percentual
tabela_intermediaria_3["variacao_anual"] = (
    tabela_intermediaria_3["valor"]
    / tabela_intermediaria_3["valor_anterior"]
    - 1
) * 100

display(tabela_intermediaria_3.tail(12))
```

```python

# Tabela Intermediária 4 - Variação em pontos percentuais

tabela_intermediaria_4 = tabela_intermediaria_3.copy()

tabela_intermediaria_4["variacao_pp"] = (
    tabela_intermediaria_4["valor"]
    - tabela_intermediaria_4["valor_anterior"]
)

tabela_intermediaria_4 = tabela_intermediaria_4[
    ["periodo", "indicador", "valor",
     "valor_anterior", "variacao_pp"]
]

display(tabela_intermediaria_4.tail(12))
```

```python

# Selecionar as 3 últimas observações
ultimos_periodos = (
    tabela_base["periodo"]
    .sort_values()
    .drop_duplicates()
    .tail(3)
)

# Transformar de formato longo para largo
tabela_final = (
    tabela_base[
        tabela_base["periodo"].isin(ultimos_periodos)
    ]
    .pivot(
        index="indicador",
        columns="periodo",
        values="valor"
    )
)

# Adicionar as comparações anuais mais recentes
comparacoes = (
    tabela_intermediaria_4
    .sort_values("periodo")
    .groupby("indicador")
    .tail(1)
    .set_index("indicador")
)

tabela_final["Selic 12 meses antes"] = (
    comparacoes["valor_anterior"]
)

tabela_final["Variação anual (p.p.)"] = (
    comparacoes["variacao_pp"]
)

# Organizar os nomes das colunas
tabela_final.columns = [
    c.strftime("%m/%Y") if isinstance(c, pd.Timestamp)
    else c
    for c in tabela_final.columns
]

display(tabela_final)
```

```python

# Criar uma versão organizada da tabela final
tabela_apresentacao = tabela_final.copy()

# Adicionar a variação percentual anual
variacao_percentual = (
    tabela_intermediaria_3
    .sort_values("periodo")
    .groupby("indicador")
    .tail(1)
    .set_index("indicador")["variacao_anual"]
)

tabela_apresentacao["Variação anual (%)"] = variacao_percentual

# Arredondar os valores
tabela_apresentacao = tabela_apresentacao.round(2)

# Exibir a tabela
print("GRUPO 4 - POLÍTICA MONETÁRIA")
print("Taxa Selic Meta - Banco Central do Brasil")
print("Série SGS: 432")
print("Valores em % ao ano; diferenças em pontos percentuais")

display(tabela_apresentacao)
```

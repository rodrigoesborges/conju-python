```r
# 1. Instalar e carregar o pacote (remova o '#' abaixo se ainda não tiver instalado)
# install.packages("ggplot2")
library(ggplot2)

# 2. Preparando os dados (Formato Tidy)
dados_informalidade <- data.frame(
  Trimestre = c("1º trimestre 2026", "2º trimestre 2026"),
  Taxa_Informalidade = c(37.3, 37.4),
  Var_Trimestre_Anterior = c(-0.2, 0.1)
)

# 3. Criando o gráfico com ggplot2
grafico <- ggplot(dados_informalidade, aes(x = Trimestre, y = Taxa_Informalidade, fill = Trimestre)) +
  # Desenha as barras
  geom_col(width = 0.5, show.legend = FALSE) +

  # Adiciona os rótulos com os valores exatos acima das barras
  geom_text(
    aes(label = paste0(Taxa_Informalidade, "%")),
    vjust = -0.6,
    size = 5,
    fontface = "bold",
    color = "#2c3e50"
  ) +

  # Ajusta o limite do eixo Y para dar respiro ao texto acima da barra
  scale_y_continuous(limits = c(0, 45), breaks = seq(0, 40, 10)) +

  # Define uma paleta de cores profissional (tons de azul)
  scale_fill_manual(values = c("1º trimestre 2026" = "#3498db", "2º trimestre 2026" = "#1f618d")) +

  # Títulos, subtítulos e fonte dos dados
  labs(
    title = "Taxa de Informalidade no Brasil (2026)",
    subtitle = "Pessoas de 14 anos ou mais ocupadas na semana de referência (%)",
    x = NULL,
    y = "Taxa de Informalidade (%)",
    caption = "Fonte: IBGE - Pesquisa Nacional por Amostra de Domicílios Contínua (Tabela 8529)"
  ) +

  # Aplica um tema limpo e moderno
  theme_minimal(base_size = 13) +
  theme(
    plot.title = element_text(face = "bold", size = 16, color = "#1a252f"),
    plot.subtitle = element_text(size = 11, color = "#5d6d7e", margin = margin(b = 15)),
    plot.caption = element_text(size = 9, color = "#7f8c8d", margin = margin(t = 15)),
    panel.grid.major.x = element_blank(), # Remove linhas verticais de grade
    axis.text.x = element_text(face = "bold", size = 12, color = "#2c3e50")
  )

# 4. Exibir o gráfico
print(grafico)
```

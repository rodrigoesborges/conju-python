```r
# 1. Instalar e carregar o pacote (remova o '#' abaixo se ainda não tiver instalado)
# install.packages("ggplot2")
library(ggplot2)

# 2. Preparando os dados (Formato Tidy)
dados_informalidade <- data.frame(
  Trimestre = c("1º trimestre 2026", "2º trimestre 2026"),
  Taxa_Informalidade = c(37.3, 37.4),
  Var_Trimestre_Anterior = c(-0.2, 0.1)
)

# 3. Criando o gráfico com ggplot2
grafico <- ggplot(dados_informalidade, aes(x = Trimestre, y = Taxa_Informalidade, fill = Trimestre)) +
  # Desenha as barras
  geom_col(width = 0.5, show.legend = FALSE) +

  # Adiciona os rótulos com os valores exatos acima das barras
  geom_text(
    aes(label = paste0(Taxa_Informalidade, "%")),
    vjust = -0.6,
    size = 5,
    fontface = "bold",
    color = "#2c3e50"
  ) +

  # Ajusta o limite do eixo Y para dar respiro ao texto acima da barra
  scale_y_continuous(limits = c(0, 45), breaks = seq(0, 40, 10)) +

  # Define uma paleta de cores profissional (tons de azul)
  scale_fill_manual(values = c("1º trimestre 2026" = "#3498db", "2º trimestre 2026" = "#1f618d")) +

  # Títulos, subtítulos e fonte dos dados
  labs(
    title = "Taxa de Informalidade no Brasil (2026)",
    subtitle = "Pessoas de 14 anos ou mais ocupadas na semana de referência (%)",
    x = NULL,
    y = "Taxa de Informalidade (%)",
    caption = "Fonte: IBGE - Pesquisa Nacional por Amostra de Domicílios Contínua (Tabela 8529)"
  ) +

  # Aplica um tema limpo e moderno
  theme_minimal(base_size = 13) +
  theme(
    plot.title = element_text(face = "bold", size = 16, color = "#1a252f"),
    plot.subtitle = element_text(size = 11, color = "#5d6d7e", margin = margin(b = 15)),
    plot.caption = element_text(size = 9, color = "#7f8c8d", margin = margin(t = 15)),
    panel.grid.major.x = element_blank(), # Remove linhas verticais de grade
    axis.text.x = element_text(face = "bold", size = 12, color = "#2c3e50")
  )

# 4. Exibir o gráfico
print(grafico)
```

```r
# 1. Instalar o pacote patchwork se não estiver instalado
if (!requireNamespace("patchwork", quietly = TRUE)) {
  install.packages("patchwork")
}
library(ggplot2)
library(patchwork)

# 2. Dados estruturados
dados_informalidade <- data.frame(
  Trimestre = c("1º Trimestre\n2026", "2º Trimestre\n2026"),
  Taxa_Informalidade = c(37.3, 37.4),
  Coeficiente_Variacao = c(0.5, 0.5),
  Var_Trimestre_Anterior = c(-0.2, 0.1)
)

# Gráfico A: Taxa de Informalidade (Barras)
g1 <- ggplot(dados_informalidade, aes(x = Trimestre, y = Taxa_Informalidade, fill = Trimestre)) +
  geom_col(width = 0.4, show.legend = FALSE) +
  geom_text(aes(label = paste0(Taxa_Informalidade, "%")), vjust = -0.5, fontface = "bold", size = 4) +
  scale_y_continuous(limits = c(0, 45)) +
  scale_fill_manual(values = c("#3498db", "#1f618d")) +
  labs(title = "Taxa de Informalidade", y = "Porcentagem (%)", x = NULL) +
  theme_minimal() +
  theme(
    plot.title = element_text(face = "bold", size = 12, color = "#2c3e50"),
    panel.grid.major.x = element_blank()
  )

# Gráfico B: Coeficiente de Variação (Linha/Pontos)
g2 <- ggplot(dados_informalidade, aes(x = Trimestre, y = Coeficiente_Variacao, group = 1)) +
  geom_line(color = "#e67e22", linewidth = 1.2) +
  geom_point(color = "#d35400", size = 3) +
  geom_text(aes(label = paste0(Coeficiente_Variacao, "%")), vjust = -1, fontface = "bold", size = 4) +
  scale_y_continuous(limits = c(0, 1)) +
  labs(title = "Coeficiente de Variação (Erro)", y = "Porcentagem (%)", x = NULL) +
  theme_minimal() +
  theme(
    plot.title = element_text(face = "bold", size = 12, color = "#2c3e50"),
    panel.grid.major.x = element_blank()
  )

# Gráfico C: Variação em Relação ao Trimestre Anterior (Barras Horizontais ou Verticais)
g3 <- ggplot(dados_informalidade, aes(x = Trimestre, y = Var_Trimestre_Anterior, fill = Var_Trimestre_Anterior > 0)) +
  geom_col(width = 0.4, show.legend = FALSE) +
  geom_text(aes(label = paste0(ifelse(Var_Trimestre_Anterior > 0, "+", ""), Var_Trimestre_Anterior, "%")),
            vjust = ifelse(dados_informalidade$Var_Trimestre_Anterior > 0, -0.5, 1.2),
            fontface = "bold", size = 4) +
  scale_y_continuous(limits = c(-0.5, 0.5)) +
  scale_fill_manual(values = c("TRUE" = "#2ecc71", "FALSE" = "#e74c3c")) + # Verde para alta, vermelho para queda
  labs(title = "Variação Trimestral", y = "Diferença p.p.", x = NULL) +
  theme_minimal() +
  theme(
    plot.title = element_text(face = "bold", size = 12, color = "#2c3e50"),
    panel.grid.major.x = element_blank()
  )

# 3. Combinando os gráficos usando o patchwork
painel <- (g1 | g2 | g3) +
  plot_annotation(
    title = "PAINEL DE INFORMALIDADE NO BRASIL (2026)",
    subtitle = "Acompanhamento dos indicadores da PNAD Contínua (Tabela 8529 - IBGE)",
    caption = "Fonte: IBGE | Nota: Letra A indica sem significância estatística na variação.",
    theme = theme(
      plot.title = element_text(size = 16, face = "bold", color = "#1a252f", hjust = 0.5),
      plot.subtitle = element_text(size = 11, color = "#7f8c8d", hjust = 0.5, margin = margin(b = 20)),
      plot.caption = element_text(size = 8, color = "#95a5a6", face = "italic")
    )
  )

# 4. Ajustar dimensões de saída do plot e exibir
options(repr.plot.width = 12, repr.plot.height = 5)
print(painel)
```
